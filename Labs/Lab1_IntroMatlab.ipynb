# 2. Theoretical aspects

## 2.1 The Matlab environment

When you start Matlab, the desktop window contains three parts that we
will use all semester:

- **Command Window** — the central pane where you type expressions and
  see the result immediately. Every command is typed after the prompt
  `>>`.
- **Workspace** — the side panel that lists all the variables that
  currently exist in memory, together with their size and value.
- **Editor** — a separate window (button “New Script” or `edit` in the
  Command Window) where you write multi-line collections of commands,
  called *scripts*. All labs are organized as scripts, and you are
  encouraged to work directly in the Editor and run the whole script (or
  a section) with the `Run` button or the `F5` key.

Whenever you type an expression in the Command Window, Matlab
*evaluates* it and prints the result. This is called **interactive
mode**.

Some useful general commands:

- `clc` — clears the Command Window;
- `clear` — deletes all variables from the Workspace;
- `doc functionname` or `help functionname` — opens the documentation
  page of a function (e.g. `help sin`). There is no way to memorize all
  Matlab functions; you search through documentation to discover them.

## 2.2 Basic math: the Command Window as a calculator

Type the following expressions, one by one, and observe the results:

``` {matlab}
2 + 2

10 / 4

2^10

5 * (3 - 8)

sqrt(2)

sin(pi / 2)
```

Matlab has the usual arithmetic operators `+`, `-`, `*`, `/` and `^`
(power), and the usual functions `sin`, `cos`, `tan`, `exp`, `log`,
`sqrt`, `abs`, `round`, … Parentheses `()` are used for grouping,
exactly as in mathematics.

Unlike a common calculator, Matlab knows the special constants `pi`,
`Inf` (infinity) and `NaN` (Not a Number, e.g. the result of `0/0`). Try
it:

``` {matlab}
pi

1/0

0/0
```

**Note**: Matlab convention uses `log(x)` for the natural logarithm
($\ln x$). The base-10 logarithm is `log10(x)`.

## 2.3 Variables

A variable is a name (starting with a letter) that stores a value.
Assign a value with the `=` operator:

``` {matlab}
a = 3

b = 2*a^2 + 5

name = "DEDP"
```

The value of the variable `a` can be used in any later expression.
Notice that after each assignment the result is printed in the Command
Window, and the new variable appears in the Workspace panel.

If you put a semicolon `;` at the end of an assignment, Matlab still
stores the variable, but **does not print** the result:

``` {matlab}
a = 3;
b = 2*a^2 + 5;
```

This is the standard style inside scripts: variables are created
silently, and we print only what we want to see.

Inspect your variables in the Workspace panel, or with the commands
`who` (list names only) and `whos` (names, sizes, and memory):

``` {matlab}
whos
```

## 2.4 Functions and keywords

A **function** is a named operation that takes input values (arguments,
in parentheses) and returns an output. Matlab has hundreds of built-in
functions for any mathematical operation: we already met `sin`, `cos`,
`sqrt`, `exp`, `log`, `abs`, `round`, `size`, `length`, `numel`, …

A function can have several arguments and several outputs. The outputs
are wrapped in square brackets:

``` {matlab}
% One argument, one output
x = sin(pi/2)

% Two arguments, one output
m = mod(17, 5)      % remainder of 17 divided by 5: 2

% One argument, two outputs: the maximum value and its position
v = [3, 9, 1, 7];
[mx, idx] = max(v)      % mx = 9, idx = 2
```

Try `help mod` and `help max` to see how the built-in help page looks.
The page always shows the calling syntax of the function, i.e. how many
arguments it accepts and in what order.

A **keyword** is a word that has a fixed, built-in meaning in the Matlab
language and cannot be used as a variable name. The most common
keywords, which will be used constantly in the following labs, are:

- `if`, `else`, `elseif`, `end` — conditional execution and the end of a
  block;
- `for`, `while` — loops (repeated execution);
- `function` — the definition of a user function;
- `return`, `break`, `continue` — control flow inside functions and
  loops;
- `true`, `false` — the logical values.

For example, the conditional structure has the form:

``` {matlab}
x = 3;
if x > 0
    disp('x is positive')
else
    disp('x is not positive')
end
```

Loops repeat a block of commands. A `for` loop repeats a fixed number of
times, iterating over the elements of a vector:

``` {matlab}
% Sum the numbers from 1 to 10
s = 0;
for k = 1:10
    s = s + k;
end
disp(s)     % 55
```

A `while` loop repeats as long as a condition is true (the condition is
re-evaluated before every repetition):

``` {matlab}
% First power of 2 that exceeds 100
p = 1;
while p <= 100
    p = 2*p;
end
disp(p)     % 128
```

Notice that `for`, `while`, `if` and `function` all delimit their block
with the keyword `end` (or `end`,`else`, alternately for `if`): this is
a general rule of the Matlab language.

A **user-defined function** is a function that we write ourselves, with
the keyword `function`. A user function is normally written in its own
file, with the same name as the function (e.g. a function called
`myfunc` is stored in the file `myfunc.m`). As an example, the following
file `mymean.m` defines a function that computes the mean of the
elements of a vector:

``` {matlab}
% File: mymean.m

function m = mymean(x)
% MYMEAN  Mean of the elements of the vector x
    m = sum(x) / length(x);
end
```

Once the file is saved, the function is called from the Command Window
or from any script, exactly like a built-in function:

``` {matlab}
v = [1, 2, 3, 4, 5];
m = mymean(v)       % 3
```

The first line of a function is called the **signature**: it declares
the output arguments (here `m`), the function name (`mymean`), and the
input arguments (here `x`). The comment line below it is the so-called
H1 help line, which is what `help mymean` displays in the Command
Window. The inputs `x` and the output `m` are variables local to the
function: they do not appear in the Workspace, and do not affect the
rest of the script.

*Note*: the `disp()` function prints a value or a text in the Command
Window, and `%` starts a comment (everything after `%` on a line is
ignored by Matlab). You will see both in every script from now on.

## 2.5 Vectors and matrices

Vectors are the most important Matlab data type for this course: **a
signal is a vector**, and **an image is a matrix**. Every laboratory
will manipulate vectors, so this section is the foundation of everything
that follows.

### Creating vectors and matrices

A **row vector** is created with square brackets `[]`, the elements
separated by spaces or commas:

``` {matlab}
v = [1, 2, 3, 4, 5]

w = [1 2 3 4 5]
```

A **column vector** separates the elements with semicolons:

``` {matlab}
v = [1; 2; 3; 4; 5]
```

A **matrix** combines the two: rows are separated by semicolons, and the
elements of a row by spaces or commas:

``` {matlab}
M = [1, 2, 3; 4, 5, 6; 7, 8, 9]
```

*Question*: what is the difference between the row vector `[1, 2, 3]`
and the column vector `[1; 2; 3]`?

### The colon operator — regularly spaced sequences

The colon operator `a:step:b` creates a vector with the values
$a, a+step, a+2\cdot step, ...$ up to `b`. The middle value is the
**step**:

``` {matlab}
x = 0:1:10

x = 0:0.1:1

x = 10:-2:0
```

Note that a negative step generates a decreasing sequence. If the step
is omitted, it is 1: `0:10` is the same as `0:1:10`. The last value of
the vector never exceeds `b` (for a positive step), so the number of
elements is not necessarily fixed.

The colon operator is the standard way to build a **time axis** for
generating signals:

``` {matlab}
t = 0:0.01:1;        % one second of time, sampled every 0.01s (100 samples)
x = sin(2*pi*5*t);   % a 5 Hz sine wave, sampled on the axis t
plot(t, x)
```

### linspace — a given number of points between two values

The function `linspace(a, b, n)` creates a vector with exactly $n$
regularly spaced values between `a` and `b`, **including both
endpoints**:

``` {matlab}
x = linspace(0, 1, 5)      % 5 points: 0, 0.25, 0.5, 0.75, 1

x = linspace(0, 1, 100)    % 100 points, step 1/99
```

The difference from the colon operator:

- `a:step:b` — we fix the **step** and let the number of points follow;
- `linspace(a, b, n)` — we fix the **number of points** and let the step
  follow (the step is always $(b-a)/(n-1)$).
- `linspace(a, b, n)` always contains the endpoints `a` and `b`, while
  the colon operator may or may not reach exactly `b`.

Both produce the same kind of regular sequence, and both are used for
generating time axes and sampling signals, so it is good practice to be
comfortable with both.

### Sizes and dimensions

The size of an array is obtained with `size()`, and the total number of
elements with `numel()`:

``` {matlab}
M = [1, 2, 3; 4, 5, 6; 7, 8, 9];
size(M)

numel(M)

length(M)   % size along the largest dimension
```

A scalar is a $1 \times 1$ matrix: `size(5)` returns `1 1`.

## 2.6 Operations on vectors and matrices

Matlab treats every variable as a matrix. This gives two completely
different types of operations, and *confusing them is the most common
beginner mistake*:

- **matrix (linear algebra) operations**: `*`, `/`, `^` — the usual
  matrix product, inverse, power;
- **element-wise operations**: `.*`, `./`, `.^` — the same operation
  applied separately to each element (notice the dot!);

For scalars (single numbers) the two are identical, which is why we
could write `2^10` above. For vectors and matrices they are different:

``` {matlab}
a = [1, 2, 3];
b = [4, 5, 6];

a .* b        % element-wise product: [4, 10, 18]

a .^ 2        % square each element: [1, 4, 9]

a * b         % ERROR: matrix product of 1x3 with 1x3 is undefined
```

The element-wise forms are the ones used for signals: they apply an
operation to every sample of a signal.

## 2.7 Indexing: accessing elements of a vector

The elements of a vector are accessed with parentheses `()`. Matlab
indexes from **1**:

``` {matlab}
v = [10, 20, 30, 40, 50];
v(1)      % the first element: 10

v(3)      % the third element: 30

v(2:4)    % elements 2, 3, 4

v(end)    % the last element
```

For a matrix we use two indices, row and column:

``` {matlab}
M = [1, 2, 3; 4, 5, 6];
M(1, 2)     % row 1, column 2: 2

M(2, :)     % everything on row 2: 4 5 6

M(:, 3)     % everything on column 3: 3 6
```

The colon `:` alone means “all elements along that dimension”.

### Logical indexing

A vector can also be indexed with a *logical* vector (a vector of
`true`/`false` values), which selects only the elements where the
logical value is `true`. This gives a very concise way to filter data:

``` {matlab}
v = [10, 3, 25, 8, 30, 1];
sel = v > 8;        % logical vector: which elements exceed 8?
v(sel)              % the elements themselves: 10 25 30

v(v > 8)            % same thing, written in one line
```

Logical indexing is the tool we will use later for finding outliers,
counting false alarms, extracting image regions, etc.

## 2.8 Basic graphics

The function `plot()` draws a vector as a sequence of samples, connected
by a line:

``` {matlab}
t = 0:0.01:1;
x = sin(2*pi*3*t);
plot(t, x)
```

When you run this, a figure window opens. A few things to remember:

- the `x` axis of the plot **is not automatic**: it is given by the
  first argument (here the time axis `t`). `plot(x)` without a time axis
  would just plot the samples against their index;
- the figure can be annotated with `xlabel`, `ylabel`, `title`, and
  `grid on`;
- `stem(t, x)` is another useful plot for discrete signals, which draws
  a marker for each sample instead of a line. Use `stem` when you want
  to see the individual *samples* of a signal, and `plot` when you want
  to see the *shape*.

``` {matlab}
t = 0:0.01:1;
x = sin(2*pi*3*t);
plot(t, x)
xlabel('Time (s)')
ylabel('Amplitude')
title('A 3 Hz sine wave')
grid on
```
